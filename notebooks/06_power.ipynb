## Power
The true δ must sit z₁₋β·SE beyond the cut-off z₁₋α/₂·SE so δ = z₁₋α/₂·SE + z₁₋β·SE so $$\delta = (z_{1-\alpha/2} + z_{1-\beta})\cdot SE$$ -- (1)

For proportions and assuming users per arm are equal, $$SE = \sqrt{\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}}$$ -- (2)

use value of SE from (2) into (1) we have $$\delta = (z_{1-\alpha/2} + z_{1-\beta})\cdot \sqrt{\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}}$$

squaring both sides, $$\delta^2 = (z_{1-\alpha/2} + z_{1-\beta})^2\cdot ({\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}})$$

so $${n} = (z_{1-\alpha/2} + z_{1-\beta})^2\cdot \frac{({p_1(1-p_1)} + {p_2(1-p_2)})}{\delta^2}$$
where $$\delta = {p_2} - {p_1}$$

## Sample size derivation (two proportions)
1. The rejection cut-off $(c)$ comes from $\alpha$ (the $H_{0}$ curve) $$c = z_{1-\alpha}\cdot SE$$
2. Power $(1 - \beta)$ comes from area under the curve of $H_{1}$ beyond the rejection cut-off $(c)$. The other tail (D < −c) is negligible when $\delta > 0$, so ignore it. $$(1 - \beta) = P(D > c | H_{1})$$
    1. subtracting $\delta$ from both sides and dividing by SE, $\frac{D - \delta}{SE} > \frac{c-\delta}{SE}$
    2. since power = P(Z > x) = 1 − β, so x has β of the area to its left, so that makes $\frac{c-\delta}{SE} = z_{\beta} = -z_{1-\beta}$
    3. using value of $(c)$ from step 1, $$\delta = z_{1-\alpha}\cdot SE + z_{1-\beta}\cdot {SE}$$
3. This means true $\delta$ must sit $z_{1-\beta}\cdot {SE}$ beyond the cut-off $z_{1-\alpha}\cdot SE$
4. For proportions and assuming users per arm are equal $$SE = \sqrt{\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}}$$
5. use value of SE from step 4 into step 2
    1. we get $\delta = (z_{1-\alpha/2} + z_{1-\beta})\cdot \sqrt{\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}}$
    2. squaring both sides, $\delta^2 = (z_{1-\alpha/2} + z_{1-\beta})^2\cdot ({\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}})$
    3. at last $${n} = (z_{1-\alpha/2} + z_{1-\beta})^2\cdot \frac{({p_1(1-p_1)} + {p_2(1-p_2)})}{\delta^2}$$
where $\delta = {p_2} - {p_1}$

### Sample size derivation (means)
4. For means, assuming users per arm are equal and standard deviation is also same
    1. $SE = \sqrt{\frac{\sigma^2}{n} + \frac{\sigma^2}{n}} = \sigma\sqrt{\frac{2}{n}}$
5. use value of SE from step 4 into step 2
    1. we get $\delta = (z_{1-\alpha/2} + z_{1-\beta})\cdot \sigma\sqrt{\frac{2}{n}}$
    2. squaring both sides, $\delta^2 = (z_{1-\alpha/2} + z_{1-\beta})^2\cdot \sigma^2\cdot\frac{2}{n}$
    3. at last $${n} = \frac{(z_{1-\alpha/2} + z_{1-\beta})^2\cdot \sigma^2}{\delta^2}$$

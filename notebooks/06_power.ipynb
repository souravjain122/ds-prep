## Sample size derivations

### Sample size derivation (two proportions) (Equal Arms)
1. The rejection cut-off $(c)$ comes from $\alpha$ (the $H_{0}$ curve) $$c = z_{1-\frac{\alpha}{2}}\cdot SE$$
2. Power $(1 - \beta)$ comes from area under the curve of $H_{1}$ beyond the rejection cut-off $(c)$. The other tail (D < −c) is negligible when $\delta > 0$, so ignore it. $$(1 - \beta) = P(D > c | H_{1})$$
    1. subtracting $\delta$ from both sides and dividing by $SE$, $\frac{D - \delta}{SE} > \frac{c-\delta}{SE}$
    2. since power = P(Z > x) = 1 − β, so x has β of the area to its left, so that makes $\frac{c-\delta}{SE} = z_{\beta} = -z_{1-\beta}$
    3. using value of $(c)$ from step 1, $$\delta = z_{1-\frac{\alpha}{2}}\cdot SE + z_{1-\beta}\cdot {SE}$$
3. This means true $\delta$ must sit $z_{1-\beta}\cdot {SE}$ beyond the cut-off $z_{1-\frac{\alpha}{2}}\cdot SE$
4. For proportions and assuming users per arm are equal $$SE = \sqrt{\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}}$$
5. use value of SE from step 4 into step 2
    1. we get $\delta = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})\cdot \sqrt{\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}}$
    2. squaring both sides, $\delta^2 = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2\cdot ({\frac{p_1(1-p_1)}{n} + \frac{p_2(1-p_2)}{n}})$
    3. at last $${n} = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2\cdot \frac{({p_1(1-p_1)} + {p_2(1-p_2)})}{\delta^2}$$ where $\delta = {p_2} - {p_1}$

### Sample size derivation (two proportions) (Unequal Arms)
4. For proportions and $n_{2} = k\cdot n_{1}$ $$SE = \sqrt{\frac{p_1(1-p_1)}{n_{1}} + \frac{p_2(1-p_2)}{k\cdot n_{1}}}$$
5. use value of SE from step 4 into step 2
    1. we get $\delta = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})\cdot \sqrt{\frac{p_1(1-p_1)}{n_{1}} + \frac{p_2(1-p_2)}{k\cdot n_{1}}}$
    2. squaring both sides, $\delta^2 = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2\cdot ({\frac{p_1(1-p_1)}{n_{1}} + \frac{p_2(1-p_2)}{k\cdot n_{1}}})$
    3. at last $$n_{1} = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2\cdot \frac{({k\cdot p_1(1-p_1)} + {p_2(1-p_2)})}{k\cdot \delta^2}$$ where $\delta = {p_2} - {p_1}$

### Sample size derivation (two proportions) (Unequal Arms) (Pooled SE under H₀)
1. The rejection cut-off $(c)$ comes from $\alpha$ (the $H_{0}$ curve) $$c = z_{1-\frac{\alpha}{2}}\cdot SE_{0}$$
2. Power $(1 - \beta)$ comes from area under the curve of $H_{1}$ beyond the rejection cut-off $(c)$. The other tail (D < −c) is negligible when $\delta > 0$, so ignore it. $$(1 - \beta) = P(D > c | H_{1})$$
    1. subtracting $\delta$ from both sides and dividing by $SE_{1}$, $\frac{D - \delta}{SE_{1}} > \frac{c-\delta}{SE_{1}}$
    2. since power = P(Z > x) = 1 − β, so x has β of the area to its left, so that makes $\frac{c-\delta}{SE_{1}} = z_{\beta} = -z_{1-\beta}$
    3. using value of $(c)$ from step 1, $$\delta = z_{1-\frac{\alpha}{2}}\cdot SE_{0} + z_{1-\beta}\cdot {SE_{1}}$$
3. This means true $\delta$ must sit $z_{1-\beta}\cdot {SE_{1}}$ beyond the cut-off $z_{1-\frac{\alpha}{2}}\cdot SE_{0}$
4. For proportions and $n_{2} = k\cdot n_{1}$ $$SE_{0} = \sqrt{\frac{p(1-p)}{n_{1}}(\frac{1+k}{k})}$$ where $p = (p_{1} + kp_{2})/(k+1)$ and $$SE_{1} = \sqrt{\frac{p_1(1-p_1)}{n_{1}} + \frac{p_2(1-p_2)}{k\cdot n_{1}}}$$
5. use value of SEs from step 4 into step 2
    1. we get $\delta = z_{1-\frac{\alpha}{2}}\sqrt{\frac{p(1-p)}{n_{1}}(\frac{1+k}{k})} + z_{1-\beta}\cdot \sqrt{\frac{p_1(1-p_1)}{n_{1}} + \frac{p_2(1-p_2)}{k\cdot n_{1}}}$
    2. multiplying both sides by $\sqrt{n_{1}}$ and dividing by $\delta$ and squaring them
    3. at last $$n_{1} = (z_{1-\frac{\alpha}{2}/2}\sqrt{{p(1-p)}(\frac{1+k}{k})} + z_{1-\beta}\cdot \sqrt{{p_1(1-p_1)} + \frac{p_2(1-p_2)}{k}})^2/\delta^2$$

### Sample size derivation (means) (Equal Arms)
4. For means, assuming users per arm are equal and standard deviation is also same
    1. $SE = \sqrt{\frac{\sigma^2}{n} + \frac{\sigma^2}{n}} = \sigma\sqrt{\frac{2}{n}}$
5. use value of SE from step 4 into step 2
    1. we get $\delta = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})\cdot \sigma\sqrt{\frac{2}{n}}$
    2. squaring both sides, $\delta^2 = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2\cdot \sigma^2\cdot\frac{2}{n}$
    3. at last $${n} = {2}\sigma^2\frac{(z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2}{\delta^2}$$

### Sample size derivation (means) (Unequal Arms)
4. For means, and standard deviation same
    1. $SE = \sqrt{\frac{\sigma^2}{n_{1}} + \frac{\sigma^2}{k\cdot n_{1}}} = \sigma\sqrt{1 + \frac{1}{k\cdot n}}$
5. use value of SE from step 4 into step 2
    1. we get $\delta = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})\cdot \sigma\sqrt{1 + \frac{1}{k\cdot n}}$
    2. squaring both sides, $\delta^2 = (z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2\cdot \sigma^2({1 + \frac{1}{k\cdot n}})$
    3. at last $${n} = \sigma^2\frac{(z_{1-\frac{\alpha}{2}/2} + z_{1-\beta})^2}{\delta^2}(\frac{k+1}{k})$$

### Validating if CI contains power for given sample size

```python
rng = np.random.default_rng(0)

x1s = rng.binomial(n=47036, p=0.120, size=2000)
x2s = rng.binomial(n=47036, p=0.126, size=2000)

results = []

for x1, x2 in zip(x1s, x2s):
    res = two_prop_ztest(x1, 47036, x2, 47036)
    results.append(res["p_value"])

pvals = np.array(results)
power = (pvals < 0.05).mean()
print(f"Power: {power}")
```

```python
n_sims = len(pvals)
se = np.sqrt(power * (1 - power) / n_sims)
lo, hi = power - 1.96 * se, power + 1.96 * se
print(f"Power = {power:.3f}, 95% CI ({lo:.3f}, {hi:.3f})")
```

1. as per formulae, n ∝ 1/δ² so if we double the MDE, n becomes one-fourth
2. increasing the power from 0.8 to 0.9 increases $z_{1-\beta}$ from 0.84 to 1.28, predicting increase in n through just power is difficult without calculation
3. for simplicity using approximation of p2 = p1, we can say for a 90/10 split, total number of users becomes roughly increases by 3 times
